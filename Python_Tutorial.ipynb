Hopefully you are more comfortable with the idea of loops now. That means that we get to move onto the second type of loop, the `for` loop. We mainly use this loop in the form `for i in range(a, b):` which means to set a variable `i` to `a`, then increase it by one whenever the code inside runs. Once `i` reaches `b - 1`, the loop runs once more, then terminates. This means that during the last run of the loop, `i` is `b - 1`.

If we want `i` to range from `0` to a number instead, we can use the shorthand `for i in range(n):`. If you want `i` to increment by some other distance `d` every time the loop iterates, use `for i in range(a, b, d):`.

Let's see the for loop in action!

````python
for i in range(3, 10): # Try changing the parameters of the range.
  print("The variable i is now " + str(i) + ".")
print("The loop has now ended. The variable i is now " + str(i) + ".")
````

You might notice that we can actually implement a for loop as a while loop as shown.

````python
# This is the equivalent of `for i in range(a, b):`
a = 3
b = 10

i = a
while (i < b):
  print("i is now " + str(i) + " ")
  i += 1
i -= 1
print("The loop has terminated. In the last iteration, the value of i was " + str(i) + ".")
````

Now, what if we want to exit out of a loop before the exit condition is reached?

Turns out it is possible to do this using the **break** keyword. When the program reaches this keyword inside a loop, the program immediately exits out of the loop. This works for both for loops and while loops.

Let's say we want to loop from 3 until 10, but we want to exit once we reach 6. We can do that with the following code.

````python
for i in range(3, 10):
  print("The current value of i is: " + str(i))
  if i == 6:
    break
````

# 4. Lists

Let's say you wanted to write down a to-do list. Once you get one thing done, Python will remind you of the next thing to do! Here's an example of 3 items below.

````python
firstItem = "Do Homework"
secondItem = "Practice 40 Hours"
thirdItem = "Conquer the World"

input("Have you done: " + firstItem + "? ")
print("Nice work!")
input("Have you done: " + secondItem + "? ")
print("Nice work!")
input("Have you done: " + thirdItem + "? ")
print("oh wow")
````

You might notice that if we had more items, this would have become extremely redundant. Well, we can sort of remove some redundancy by using a loop:

````python
for i in range(3):
  if i == 0:
    item = "Do Homework"
  elif i == 1:
    item = "Practice 40 Hours"
  elif i == 2:
    item = "Conquer the World"

  input("Have you done: " + item + "? ")
  if i != 2:
    print("Nice work!")
  else:
    print("oh wow")
````

Ok, it's a bit better now, but that first part will become a lot more lengthy the more items we add to the list. We want a simple way to store all of these items so that they can easily be refered to their place in the list. So, we use a Python... List.

Lists are, well, lists of objects (numbers, booleans, and strings) which can be assigned to variables much like other datatypes can. We can initialize a list by enclosing the objects, separated by commas, in square braces:

```
toDoList = ["Do Homework", "Practice 40 Hours", "Conquer the World"]
```

Now, how do we get these items in the list? We do this with `toDoList[index]`, where `index` is the placement of the object in the list, starting from `0`. For example, if we wanted the first item in the list ("Do Homework"), we would use `toDoList[0]`. Similarly, if we wanted the last item in the list ("Conquer the World"), we would use `toDoList[2]`. Let's see our improved code below!

````python
toDoList = ["Do Homework", "Practice 40 Hours", "Conquer the World"]

for i in range(len(toDoList)): # len(list) is a function that returns the length of the list as an integer
  input("Have you done: " + toDoList[i] + "? ")
  if i != 2:
    print("Nice work!")
  else:
    print("oh wow")
````

Our code has become so much cleaner with the addition of lists! You can start to see how loops play well with lists, since loops can be used to **iterate** over the elements of the list.

There is actually a final type of loop we did not cover in the loops section, since a list is required to use this loop. This is the **for each** loop, and it iterates through the elements of the list without using the index. This is done with the (quite self-explanatory) `for item in list` syntax:

````python
toDoList = ["Do Homework", "Practice 40 Hours", "Conquer the World"]
print("Lots of stuff to do today!")
for item in toDoList:
  print("- " + item)
````

Now, remember the datatypes we had earlier? We can also make our own objects, although they aren't exactly datatypes.

**Objects** are instances of **classes**, and have some properties such as their own variables and functions. See an example below.

````python
class Computer:
  def __init__(self, name, width, height, backlit):
    self.name = name # string
    self.width = width # float
    self.height = height # float
    self.backlit = backlit # boolean

  def about(self):
    print("This is Computer " + self.name + ".")
    print("It has a width of " + str(self.width) + " inches and a height of " + str(self.height) + " inches.")
    if self.backlit:
      print("This computer has a backlit keyboard.")
    else:
      print("This computer does not have a backlit keyboard.")

  def rickroll(self):
    print("Turning on speakers to 100%...")
    print("Never gonna- ok i'll stop")

myComputer = Computer("myComputer", 8, 11, True)
myComputer.about()
print()

myComputer.rickroll()

# Assign your own computer to a variable, and call its about function.
# (Rickroll function is encouraged but not mandatory.)
````

Our new class, `Computer`, has three functions and four **instance variables**. We made an **instance** of this class, myComputer, which is called an **object**.

Our first function is the `__init__` function, and this function is especially used when making a new object of this class. This function even has a special name, the **constructor** function. So when `Computer("name", width, height, backlit)` is called, these arguments are sent to the `__init__` function. Then a new object is made with these arguments.

Now, the next two functions can be called with the format `obj.function()`. The `self` argument, which is the object itself, is added implicitly. Other functions that have other arguments need those arguments to be set.
